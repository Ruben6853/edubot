```python
import numpy as np

def rot_x(roll: float) -> np.ndarray:
    c = np.cos(roll)
    s = np.sin(roll)
    return np.array([
        [1.0, 0.0, 0.0],
        [0.0, c,  -s ],
        [0.0, s,   c ]
    ])

def rot_y(pitch: float) -> np.ndarray:
    c = np.cos(pitch)
    s = np.sin(pitch)
    return np.array([
        [ c, 0.0,  s],
        [0.0, 1.0, 0.0],
        [-s, 0.0,  c]
    ])

def pose_to_rotation(pitch: float, roll: float) -> np.ndarray:
    """
    Construct desired end-effector rotation matrix from pitch and roll.
    Assumption: yaw = 0, so R = Ry(pitch) @ Rx(roll)
    """
    return rot_y(pitch) @ rot_x(roll)

def pose_to_transform(x: float, y: float, z: float, pitch: float, roll: float) -> np.ndarray:
    """
    Build a 4x4 homogeneous transform from [x, y, z, pitch, roll].
    """
    R = pose_to_rotation(pitch, roll)
    T = np.eye(4)
    T[:3, :3] = R
    T[:3, 3] = np.array([x, y, z])
    return T
```
